```python
global_temp = pd.read_csv("data/processed/weather/global_temperature_dataset.csv")
global_temp.head()
```

```python
global_temp.info()
```

```python
global_temp = global_temp.drop(columns=["data_source", "year", "month", "month_name", "day"], axis=1)
```

```python
global_temp.columns 
```

```python
global_temp.describe()
```

```python
global_temp["date"] = pd.to_datetime(global_temp["date"])
```

```python
global_temp.head()
```

```python
base_df = global_temp[
    [
        "city",
        "country",
        "continent",
        "latitude",
        "longitude",
        "date",
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "temperature_anomaly"
    ]
].copy()

city_wise_full = base_df.pivot_table(
    index=["city", "country", "continent", "latitude", "longitude"],
    columns="date",
    values=[
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "temperature_anomaly"
    ]
)
```

```python
city_wise_full.shape
```

```python
city_wise_full.columns = [
    f"{metric}_{date.strftime('%Y-%m-%d')}"
    for metric, date in city_wise_full.columns
]
city_wise_full = city_wise_full.reset_index()
```

```python
city_wise_full.head()
```

```python
city_wise_full.to_csv("data/processed/weather/city_wise_10yr_clean.csv", index=False)
```
